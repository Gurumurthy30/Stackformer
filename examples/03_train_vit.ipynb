# Stackformer: Vision Transformer (ViT) Image Classification

This tutorial demonstrates configuring and training a Vision Transformer on image classification using `Trainer`.

```python
import torch
from torch.utils.data import DataLoader, TensorDataset

from stackformer.models.vision import ViT
from stackformer.training import TrainConfig, Trainer

# 1. Model Definition: Compact Vision Transformer for 32x32 images
model = ViT(
    img_size=32,
    patch_size=8,
    num_layers=2,
    embed_dim=32,
    num_classes=5,
    num_heads=2,
    hidden_dim=64,
)
print(f"ViT parameters: {sum(p.numel() for p in model.parameters()):,}")
```

## 2. Prepare Synthetic Image Dataset

```python
torch.manual_seed(42)
train_x = torch.randn(32, 3, 32, 32)
train_y = torch.randint(0, 5, (32,))
train_loader = DataLoader(TensorDataset(train_x, train_y), batch_size=8, shuffle=True)

val_x = torch.randn(16, 3, 32, 32)
val_y = torch.randint(0, 5, (16,))
val_loader = DataLoader(TensorDataset(val_x, val_y), batch_size=8)
```

## 3. Train with Stackformer Trainer

```python
cfg = TrainConfig(
    max_epochs=2,
    lr=1e-3,
    loss="cross_entropy",
    checkpoint_dir="./checkpoints_vit",
    eval_every_n_epochs=1,
    display="auto",
)

trainer = Trainer(model, cfg, train_loader=train_loader, eval_loader=val_loader)
state = trainer.train()
print(f"ViT training completed! Global step: {state.global_step}")
```
